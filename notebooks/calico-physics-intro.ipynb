## 2. Newton's Laws & Projectile MotionF = ma. For a projectile with no air resistance, gravity is the only force,giving a parabolic trajectory.

```python
def projectile(v0, angle_deg, g=9.8):    theta = np.radians(angle_deg)    t_flight = 2*v0*np.sin(theta)/g    t = np.linspace(0, t_flight, 100)    x = v0*np.cos(theta)*t    y = v0*np.sin(theta)*t - 0.5*g*t**2    return x, yfor angle in [30, 45, 60]:    x, y = projectile(25, angle)    plt.plot(x, y, label=f"{angle}°")plt.title("Projectile Motion at Different Launch Angles")plt.xlabel("x (m)"); plt.ylabel("y (m)"); plt.legend(); plt.grid(alpha=0.3); plt.show()
```

## 3. Energy Conservation (Pendulum)Total mechanical energy E = KE + PE should stay constant for a frictionless pendulum.We integrate the equations of motion numerically and check that E stays flat.

```python
g, L = 9.8, 1.0theta0, omega0 = np.radians(60), 0dt = 0.001theta, omega = theta0, omega0thetas, energies = [], []for _ in range(4000):    alpha = -(g/L)*np.sin(theta)    omega += alpha*dt    theta += omega*dt    KE = 0.5*(L*omega)**2    PE = g*L*(1 - np.cos(theta))    thetas.append(theta)    energies.append(KE + PE)plt.plot(energies)plt.title("Total Mechanical Energy Over Time (should be flat)")plt.xlabel("step"); plt.ylabel("Energy (per unit mass)"); plt.show()
```

## 4. Momentum & CollisionsIn an elastic collision, both momentum and kinetic energy are conserved.Below, two balls collide head-on and we verify both quantities before and after.

```python
def elastic_collision_1d(m1, v1, m2, v2):    v1f = ((m1-m2)*v1 + 2*m2*v2) / (m1+m2)    v2f = ((m2-m1)*v2 + 2*m1*v1) / (m1+m2)    return v1f, v2fm1, v1, m2, v2 = 2, 5, 1, -3v1f, v2f = elastic_collision_1d(m1, v1, m2, v2)p_before = m1*v1 + m2*v2p_after  = m1*v1f + m2*v2fKE_before = 0.5*m1*v1**2 + 0.5*m2*v2**2KE_after  = 0.5*m1*v1f**2 + 0.5*m2*v2f**2print(f"Momentum before/after: {p_before:.2f} / {p_after:.2f}")print(f"KE before/after: {KE_before:.2f} / {KE_after:.2f}")
```

## 5. Gravitation & OrbitsNewton's law of gravitation, F = GMm/r^2, produces elliptical orbits when integrated.

```python
G, M = 1.0, 1000.0pos = np.array([50.0, 0.0])vel = np.array([0.0, 4.2])dt = 0.01trail = []for _ in range(3000):    r = np.linalg.norm(pos)    acc = -G*M*pos/r**3    vel += acc*dt    pos += vel*dt    trail.append(pos.copy())trail = np.array(trail)plt.plot(trail[:,0], trail[:,1])plt.scatter([0],[0], color='orange', label='Central mass')plt.title("Orbit Under Inverse-Square Gravity")plt.axis('equal'); plt.legend(); plt.show()
```

## 9. Electric Field of a Point ChargeE = kQ/r^2, pointing radially outward from a positive charge.Below we visualize the field vectors around a single charge.

```python
k, Q = 8.99e9, 1e-9grid = np.linspace(-5, 5, 15)X, Y = np.meshgrid(grid, grid)R = np.sqrt(X**2 + Y**2) + 0.3Ex = k*Q*X/R**3Ey = k*Q*Y/R**3plt.quiver(X, Y, Ex, Ey)plt.scatter([0],[0], color='red', s=80, label='+Q')plt.title("Electric Field Around a Point Charge")plt.legend(); plt.axis('equal'); plt.show()
```

## 10. Circuits: Ohm's Law & Resistor NetworksV = IR for a single resistor. For resistors in series, resistances add;in parallel, conductances (1/R) add.

```python
def series(*rs): return sum(rs)def parallel(*rs): return 1 / sum(1/r for r in rs)R1, R2, R3 = 100, 220, 330print("Series R:", series(R1, R2, R3), "ohm")print("Parallel R:", round(parallel(R1, R2, R3), 2), "ohm")V = 9I_series = V / series(R1, R2, R3)print(f"Current through series circuit at {V}V: {I_series*1000:.2f} mA")
```

## 11. Optics: Snell's Lawn1*sin(theta1) = n2*sin(theta2). Light bends toward the normal when enteringa denser medium, and away from it when leaving one.

```python
def refraction_angle(n1, theta1_deg, n2):    theta1 = np.radians(theta1_deg)    sin_theta2 = n1*np.sin(theta1)/n2    if abs(sin_theta2) > 1:        return None  # total internal reflection    return np.degrees(np.arcsin(sin_theta2))for theta1 in [10, 30, 50, 70]:    theta2 = refraction_angle(1.0, theta1, 1.5)    print(f"Air->Glass: incidence {theta1}° -> refraction {theta2:.1f}°")
```

## 12. Special RelativityAs velocity approaches the speed of light, time dilates and length contracts,governed by the Lorentz factor gamma = 1/sqrt(1 - v^2/c^2).

```python
beta = np.linspace(0, 0.999, 200)gamma = 1/np.sqrt(1 - beta**2)plt.plot(beta, gamma)plt.title("Lorentz Factor vs Velocity (as a fraction of c)")plt.xlabel("v/c"); plt.ylabel("gamma"); plt.grid(alpha=0.3); plt.show()print(f"At v = 0.9c, time dilates by a factor of {1/np.sqrt(1-0.9**2):.2f}")
```
